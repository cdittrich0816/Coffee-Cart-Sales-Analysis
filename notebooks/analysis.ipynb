# Coffee Cart Sales Analysis

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

# Paths relative to the repo root (this notebook lives in notebooks/),
# not hardcoded to any one machine.
REPO_ROOT = Path.cwd().parent
DATA_PATH = REPO_ROOT / "data" / "raw" / "coffee_sales.csv"
OUTPUT_DIR = REPO_ROOT / "output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_SEED = 42
```

```python
df = pd.read_csv(DATA_PATH)
df["revenue"] = df["quantity"] * df["unit_price"]
df.head()
```

### Daily revenue

```python
daily = df.groupby("date")["revenue"].sum()
daily
```

### Average Daily Revenue

```python
avg_daily = daily.mean()
print(f"Average daily revenue: ${avg_daily:.2f}")
```

### Compare revenue: random 50/50 split of the days
`random_state` is set so the split (and the group means below) are the same on every run.

```python
shuffled = df.sample(frac=1, random_state=RANDOM_SEED)
half = len(shuffled) // 2
group_a = shuffled.iloc[:half]
group_b = shuffled.iloc[half:]
print(f"Group A mean revenue: {group_a['revenue'].mean():.2f}")
print(f"Group B mean revenue: {group_b['revenue'].mean():.2f}")
```

### Revenue by product

```python
by_product = df.groupby("product")["revenue"].sum()
by_product.plot(kind="bar")
plt.ylabel("Total revenue ($)")
plt.title("Revenue by product")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "revenue_by_product.png")
plt.show()
```

### Takeaway
Latte coffee brings in the most total revenue this month.
